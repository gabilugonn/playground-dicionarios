```python
def rename_dict_key(dictionary, old_key, new_key):
    print("Dicionário antes da função:", dictionary)
    if old_key in dictionary:
        dictionary[new_key] = dictionary.pop(old_key)
        print("Dicionário depois da função:", dictionary)
        return True
    else:
        print("Dicionário depois da função (chave não encontrada):")
        print(dictionary)
        return False

# Exemplo de uso:
my_dict = {"nome": "Alice", "idade": 30, "cidade": "Nova York"}
print("\n--- Teste 1: Chave existente ---")
success = rename_dict_key(my_dict, "nome", "primeiro_nome")
print("Sucesso no renomear (esperado True):", success)

print("\n--- Teste 2: Chave não existente ---")
success = rename_dict_key(my_dict, "pais", "nacionalidade")
print("Sucesso no renomear (esperado False):", success)

print("\n--- Teste 3: Outro exemplo com chave existente ---")
my_another_dict = {"produto": "Caneta", "preco": 2.50}
success = rename_dict_key(my_another_dict, "preco", "valor")
print("Sucesso no renomear (esperado True):", success)
```
